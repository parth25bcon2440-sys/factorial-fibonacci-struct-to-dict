```python
# Factorial
n = 5
fact = 1
for i in range(1, n + 1):
    fact = fact * i
print("Factorial of", n, "=", fact)

# Fibonacci series
n = 10
a, b = 0, 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
print() # New line

# Struct to dictionary
student = {
    "name": "Rahul",
    "roll": 101,
    "marks": 87.5
}

print("Name:", student["name"])
print("Roll:", student["roll"])
print("Marks:", student["marks"])
```
